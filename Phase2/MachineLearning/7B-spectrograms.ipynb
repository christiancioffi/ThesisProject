SPETTROGRAMMI

```python
from pathlib import Path
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
import shutil
import os
from PIL import Image
```

```python
NORMALIZED_AUDIO=False
normalized_str="[NORMALIZED_AUDIO]" * NORMALIZED_AUDIO
NOISE_REDUCTION=False
noise_reduction_str="[NOISE_REDUCTION]" * NOISE_REDUCTION
HIGH_PASS_FILTER=False
high_pass_filter_str="[HIGH_PASS_FILTER]" * HIGH_PASS_FILTER

AUDIO_DIR = Path(f"{normalized_str}{noise_reduction_str}{high_pass_filter_str}AudioSamplesPreprocessed")
SPECTROGRAM_DIR = Path(f"{normalized_str}{noise_reduction_str}{high_pass_filter_str}Spectrograms")

SAMPLE_RATE = 16000
FRAME_SIZE = 2048
HOP_LENGTH = 512
N_MELS = 128
```

```python
if SPECTROGRAM_DIR.exists():
    shutil.rmtree(SPECTROGRAM_DIR)  # Rimuove la cartella e tutto il contenuto

SPECTROGRAM_DIR.mkdir(parents=True) # La ricrea vuota
```

```python
audio_files = sorted(
    list(AUDIO_DIR.glob("*.wav"))
)

total_files = len(audio_files)

print(f"Trovati {total_files} file audio in {AUDIO_DIR}")
```

```python
for i, file_path in enumerate(audio_files, start=1):
    try:
        print(f"[{i}/{total_files}] Elaborazione in corso: {file_path.name}...")

        # 1. Carica il file audio
        y, _ = librosa.load(file_path, sr=SAMPLE_RATE)

        #print(f"min: {np.min(y):.4f}, max: {np.max(y):.4f}")

        # 2. Calcola il Mel Spectrogram e converti in dB
        mel_spec = librosa.feature.melspectrogram(y=y, sr=SAMPLE_RATE, n_fft=FRAME_SIZE, hop_length=HOP_LENGTH, n_mels=N_MELS)
        mel_spec_db = librosa.power_to_db(mel_spec) #1

        print(f"Lo shape dello spettrogramma è: {mel_spec_db.shape}")

        # 3. SALVATAGGIO IN FORMATO BINARIO GREZZO (.bin)
        # Forza il tipo di dati a float32 (standard per C++ e sistemi micro)
        mel_float32 = mel_spec_db.astype(np.float32)
        
        output_path = SPECTROGRAM_DIR / f"{file_path.stem}_melspec.bin"
        
        # tofile() scrive i byte grezzi della matrice sul disco, senza header o metadati
        mel_float32.tofile(output_path)

        print(f"-> Salvato con successo ({mel_float32.shape[1]}x{mel_float32.shape[0]} valori): {output_path.name}\n")

    except Exception as e:
        print(f"Errore con {file_path.name} (File {i}/{total_files}): {e}\n")

print("Elaborazione completata!")
```
